## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target / proxy: A binary proxy for declining content that may deserve review.

The proxy is defined from the observed trend_direction field:
1 = the content is currently classified as down
0 = the content is not currently classified as down

In the starter dataset, 54.2% of content items are labeled 1 and 45.8% are labeled 0.

This proxy does not prove that refreshing a page will cause its performance to improve. Instead, it provides a directional signal that can help identify content that may deserve further review.

Because the proxy is defined using trend_direction, trend_direction and trend_pct should not be used as predictive model features for this target, since that would create target leakage.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

**Unit of analysis:** One row represents one content item being evaluated for refresh opportunity.

Each row contains observable features describing the content item and its recent performance.

**One row = one content item/page.**

The target or proxy column will represent whether the content item is considered a higher refresh opportunity, or an opportunity score that can be used to rank items.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd

# Load the FlyRank starter dataset
data_url = "https://raw.githubusercontent.com/nataliehany/FlyRank-ML-Internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_url)

print("Dataset shape:", df.shape)
print("\nOne row = one content item/page")
display(df.head())
```

```python
# Show a real slice of the data for the refresh-opportunity task

lane_columns = [
    "content_id",
    "search_volume",
    "impressions_90d",
    "clicks_90d",
    "avg_position",
    "content_age_days",
    "days_since_last_update",
    "ctr",
    "trend_direction",
    "trend_pct"
]

lane_df = df[lane_columns].copy()

print("Unit of analysis: one row = one content item/page")
print("Number of content items:", len(lane_df))

display(lane_df.head(10))



```

```python
# Sketch a simple proxy target for the framing exercise
# 1 = declining content that may deserve review
# 0 = not currently declining

lane_df["refresh_opportunity_proxy"] = (
    lane_df["trend_direction"] == "down"
).astype(int)

display(
    lane_df[
        [
            "content_id",
            "impressions_90d",
            "content_age_days",
            "avg_position",
            "trend_direction",
            "refresh_opportunity_proxy"
        ]
    ].head(10)
)

print("\nProxy target distribution:")
print(lane_df["refresh_opportunity_proxy"].value_counts())

print("\nProxy target percentages:")
print(
    lane_df["refresh_opportunity_proxy"]
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could prioritize content using a simple threshold, such as refreshing every page older than a certain number of days or every page above a certain number of impressions.

However, refresh opportunity depends on several signals interacting together. For example, an older page is not necessarily a strong refresh opportunity if it has very little demand, while another page may have high impressions but weaker performance and therefore deserve more attention.

Machine learning can combine multiple signals such as content age, impressions, search demand, position, and recent performance patterns instead of relying on one manually chosen threshold.

The output supports a real content action: helping the content team decide which content items should be reviewed and potentially refreshed first.
